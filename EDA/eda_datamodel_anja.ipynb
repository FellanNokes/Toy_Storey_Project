# EDA exploration for datamodel

```python
from pathlib import Path

import duckdb

ROOT = Path.cwd().parent              # EDA/ -> Toy_Storey_Project/
DB_FILE = ROOT / "data" / "toystore.duckdb"

def read_table(table):
    with duckdb.connect(str(DB_FILE), read_only=True) as conn:
        df = conn.execute(f"SELECT * FROM raw.{table}").df()
        return df
```

```python
df_sales = read_table("sales")
df_sales_items = read_table("sales__items")
df_inventory = read_table("inventory")
df_inventory__items = read_table("inventory__items")
df_products = read_table("products")
```
