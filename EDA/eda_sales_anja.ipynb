# EDA for first showcase of dashboard

```python
from pathlib import Path
import duckdb

ROOT_PATH = Path.cwd().parent
DATA_PATH = ROOT_PATH / "data"

# Read data 
with duckdb.connect(DATA_PATH / "toystore.duckdb", read_only=True) as conn:
    df_sales = conn.execute("SELECT * FROM raw.sales").df()

with duckdb.connect(DATA_PATH / "toystore.duckdb", read_only=True) as conn:
    df_sales_items = conn.execute("SELECT * FROM raw.sales__items").df()
    
```
